```python
#Calculate the percentage of purine(A,G) in the nucleotide sequence
```

```python
sequence="ATGACCATGATTACGGATTCACTGGCCGTCGTTTTACAACGTCGTGACTGGGAAAACCCTGGCGTTACCCAACTTAATCGCCTTGCAGCACATCCCCCTTTCCAGGCCTGCGCTCGCCGCTGCCGGGCGTTTTTTGCGCAGGTGCGGCTACGGCCTGGTGATGCTGCGCCTGCGCCGCGATGTCGCGTTCGGTGCGC"
```

```python
purines = sequence.count("A") + sequence.count("G")
```

```python
percentage = (purines / len(sequence)) * 100

print("Total nucleotides:", len(sequence))
print("Number of purines (A + G):", purines)
print("Percentage of purines:", percentage, "%")
```
